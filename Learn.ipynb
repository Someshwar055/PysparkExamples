```python
import findspark
findspark.init()

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("pivot-unpivot-demo") \
    .master("local[*]") \
    .getOrCreate()

spark
```

```python
from pyspark.sql import Row
from pyspark.sql.functions import col

data = [
    Row(dept="IT",   year=2023, revenue=120),
    Row(dept="IT",   year=2024, revenue=150),
    Row(dept="HR",   year=2023, revenue=80),
    Row(dept="HR",   year=2024, revenue=90),
    Row(dept="Finance", year=2023, revenue=200),
    Row(dept="Finance", year=2024, revenue=220)
]

df = spark.createDataFrame(data)
df.show()
```
